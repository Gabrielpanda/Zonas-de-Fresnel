```python
#Este programa calcula los radios de las zonas de fresnel para frecuencias y profundidades distintas
#Para calcular dichos radios usaremos la ecuacion: R=(d*L/2)^(1/2)
import numpy as np
```

```python
# Calculamos las longitudes de onda para diferentes frecuencias:
f=[20.,45.,60.,90.,120.,250.,325.]
v=[1535,2450,3200,3800]
d=[500,1000,1500,2000]
#Funcion que calcula la longitud de onda
def longitud_onda(f,v):
    L=np.zeros(len(f), float)
    for i in range (len(f)):
        L[i]=v/f[i]
    return L
#Funcion que calcula el Radio de la zona de Fresnel
def radio(L,d):
    R=np.zeros(len(L), float)
    for i in range(len(L)):
        R[i]=((d*L[i])/2)**0.5
    return R
```

```python
#Calculo de las longitudes de onda con distintas velocidades para las frecuencias dadas
L1=longitud_onda(f,v[0])
L2=longitud_onda(f,v[1])
L3=longitud_onda(f,v[2])
L4=longitud_onda(f,v[3])
```

```python
#Funcion que calcula las zonas de Fresnel para distintas profundidade4s
def zones(L1,d):
    zone=[[],[],[],[]]
    for i in range(len(d)):
        zone[i]=radio(L1,d[i])
    return zone
```

```python
#Calculo de las zonas Fresnel con longitudes de onda y distancias variables
zone1=zones(L1,d)
zone2=zones(L2,d)
zone3=zones(L3,d)
zone4=zones(L4,d)
```

```python
for i in range(len(zone1)):
    print('Zonas de fresnel relacionadas con una distancia de:', d[i])
    print("velocidad de",v[i])
    for j in range(len(zone1[i])):
        print(zone1[i][j],"Frecuencia de:", f[j], "[Hz]")
```
